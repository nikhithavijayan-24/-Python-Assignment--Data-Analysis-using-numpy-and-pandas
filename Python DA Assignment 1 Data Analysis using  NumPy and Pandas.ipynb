```python
# Creating 1-d numpy array
temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

```python
#shape,type and size of the array
print(temperatures_w1.ndim)
print(temperatures_w1.shape)
print(temperatures_w1.size)
print(temperatures_w1.dtype)
print(type(temperatures_w1))
```

```python
# Celsius --> Fahrenheit
temperatures_fahrenheit=(temperatures_w1*9/5)+32
print(temperatures_fahrenheit)
```

```python
# Maximum, Minimum and Average Temperatures
print("Maximum:",np.max(temperatures_w1))
print()
print("Minimum:",np.min(temperatures_w1))
print()
print("Average:",np.mean(temperatures_w1))
```

```python
# Middle three elements(Temperatures)
print(temperatures_w1)
print()
print(temperatures_w1[2:5])
```

```python
# Creating 2-d array
temperatures=np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
                      [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
print(temperatures)
```

```python
# Shape, Size and types of array
print(temperatures.ndim)
print(temperatures.shape)
print(temperatures.size)
print(temperatures.dtype)
print(type(temperatures))
```

```python
# Last two elements for each array(Temperatures)
print(temperatures)
print()
print(temperatures[0,-2:])
print()
print(temperatures[1,-2:])
```

```python
# Creating panda series
mark=95, 92, 89, 85, 80
rank='Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'
marks=pd.Series(data=mark,index=rank)
print(marks)
```

```python
# Accessing the mark of Rank1
print(marks)
print()
print(marks["Rank1"])
```

```python
# First three elements using loc
print(marks.loc[["Rank1","Rank2","Rank3"]])
```

```python
# Mark for rank3 using iloc
print(marks.iloc[2])
```

```python
#Retrieving the ranks where marks >90
print(marks)
print()
print(marks[marks>90])
```

```python
# Modifying the mark of Rank1 --> 100
marks.loc["Rank1"]=100
print(marks)
```

```python
# Dropping the rank5 entry
marks.drop("Rank5",inplace=True)
print(marks)
```

```python
#Calculating the cgpa
cgpa=marks/10
print(cgpa)
```

```python
# Creating a panda dataframe
transcation={
    "TranscationID":[101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "ProductCategory":['Electronics', 'Clothing', 'Electronics', 'Furniture',
'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    "Region":['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West',
'South', 'North'],
    "Amount":[200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}
transcations=pd.DataFrame(transcation)
print(transcations)
```

```python
# Showing created pandas dataframe
print(transcations)
```

```python
# Accessing the first elements
print(transcations.head())
```

```python
#Accessing the last elements
print(transcations.tail())
```

```python
# Shape 
print(transcations.shape)
```

```python
# Column names
print(transcations.columns)
```

```python
#Data types
print(transcations.dtypes)
```

```python
# Displaying productcategory & amount column
print(transcations[["ProductCategory","Amount"]])
```

```python
# Last three elements using loc
print(transcations.loc[transcations.index[-3:]])
```

```python
# last three elements using iloc
print(transcations.iloc[-3:])
```

```python
#filtering columns - Region=North & Amount>200
print(transcations[(transcations["Region"]=="North") & (transcations["Amount"]>200)])
```

```python
# Value counts for product category
print(transcations["ProductCategory"].value_counts())
```

```python
#Unique elements in Region column
print(transcations["Region"].unique())
```

```python
# Filtering the column Region using groupby and find the average of column amount in each region
print(transcations.groupby("Region")["Amount"].mean())
```

```python
# Accessing TranscationID=102 and Amount=165 using loc
transcations.loc[transcations["TranscationID"]==102,"Amount"]=165
print(transcations)
```

```python
# New column --> Discount
transcations["Discount"]=(transcations["Amount"]*10/100)
print(transcations)
```

```python
# Removing row with transcationid=109
transcations.drop(transcations[transcations["TranscationID"] == 109].index, inplace=True, axis=0)
print(transcations)
```

```python
# Removing column--> Discount
transcations.drop("Discount",inplace=True,axis=1)
print(transcations)
```
